```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

```python
# loading csv file into pandas
df=pd.read_csv('customer_shopping_behavior.csv')
```

```python
df.head()
```

```python
df. info()
```

```python
df.describe(include='all')
```

```python
df.isnull().sum()
```

```python
# Rows having null values

df[df.isnull().any(axis=1)].head()
```

```python
# Imputed missing values in Review Rating column wih median ratings of product category
df['Review Rating']=  df.groupby('Category')['Review Rating'].transform(lambda x: x.fillna(x.median()))
df.isnull().sum()
```

```python
# Tracing duplicate entries
df[df.duplicated()] # returns duplicate rows
df.duplicated().sum() # returns sum of rows with duplicate values
```

```python
df.columns
```

```python
df.columns=df.columns.str.lower()
df.columns=df.columns.str.replace(" ","_")
df=df.rename(columns={'purchase_amount_(usd)':'purchase_amount'})
df.columns
```

```python
# Creating a new column age_group

labels = ["Young Adults","Adults","Middle-aged","Senior"]
df['age_group']=pd.qcut(df["age"],q=4,labels=labels)
df[["age","age_group"]].head(10)
```

```python
df['frequency_of_purchases'].unique()
```

```python
# create new column purchase_frequency_days

frequency_mapping = {
    'Fortnightly': 14,
    'Weekly': 7,
    'Monthly': 30,
    'Quarterly': 90,
    'Bi-Weekly': 14,
    'Annually': 365,
    'Every 3 Months': 90
}

df["purchase_frequency_days"]=df["frequency_of_purchases"].map(frequency_mapping)
```

```python
(df["discount_applied"]==df["promo_code_used"]).all()
```

```python
# Dropping promo_code_used as it is identical to discount_applied (verified above)

df=df.drop(columns=("promo_code_used"))
```

```python
# Checking for Outlier
df['purchase_amount'].hist()
plt.show()
```
